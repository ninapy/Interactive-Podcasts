```python
import altair as alt
import pandas as pd
```

```python
# data preparation

# standard deviation and mean for each group
group_stats = pd.DataFrame([
    {'group': 'FRQ', 'mean': 0.685371, 'std': 0.124625},
    {'group': 'MCQ', 'mean': 0.598950, 'std': 0.219560},
    {'group': 'None', 'mean': 0.585950, 'std': 0.303574},
])

# data for format comparison
format_data = pd.DataFrame([
    {'Group': 'Note-Takers', 'Section': 'FRQ', 'Score': 0.6007},
    {'Group': 'Note-Takers', 'Section': 'MCQ', 'Score': 0.8214},
    {'Group': 'Non Note-Takers', 'Section': 'FRQ', 'Score': 0.5555},
    {'Group': 'Non Note-Takers', 'Section': 'MCQ', 'Score': 0.5714},
])

# graph 1

base1 = alt.Chart(group_stats).encode(
    x=alt.X('group:N', sort=['FRQ', 'MCQ', 'None'], title="Group")
)

bars1 = base1.mark_bar(size=55, opacity=0.8).encode(
    y=alt.Y('mean:Q', scale=alt.Scale(domain=[0, 1]), title="Average Score"),
    color=alt.Color('group:N', scale=alt.Scale(scheme='tableau10'), legend=None),
    tooltip=['group', 'mean', 'std']
)

error1 = base1.mark_errorbar(color='black', thickness=2).encode(
    y='mean:Q',
    yError='std:Q'
)

consistency_chart = (bars1 + error1).properties(
    title="Mean Score & Standard Deviation per Group",
    width=350, height=450
).configure_title(fontSize=16, anchor='middle')

#graph 2

format_chart = alt.Chart(format_data).mark_bar(size=60).encode(
    x=alt.X('Section:N', title=None, sort=['FRQ', 'MCQ']),
    y=alt.Y('Score:Q', scale=alt.Scale(domain=[0, 1]), title="Average Score"),
    color=alt.Color('Section:N', scale=alt.Scale(scheme='tableau10'), legend=None),
    column=alt.Column('Group:N', title="Student Strategy", sort=['Note-Takers', 'Non Note-Takers']),
).properties(
    title="Efficiency of Note-Taking on Different Question Formats",
    width=180, height=350
).configure_title(
    fontSize=18,
    anchor='middle'
)

# display

consistency_chart.display()
format_chart.display()
```
